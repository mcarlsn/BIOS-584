### Q2.1. Univariate comparison (3')

<font size='4'> 
    
- Suppose that we would like to examine the utility/effect of an intervention program for patients with PTSD.
- We measure PCL5 scores at completion (`pcl5week_score.completion`) and PCL5 score at 3-month follow-up (`pcl5month_score.3_month_follow_up`). Let's assume the first score is pre-intervention and the second score is post-intervention.
- Report the summary statistics for each variable including mean, std, median, Q1, and Q3.
- Note that each patient will receive such two PCL5 scores. Use a appropriate statistical test to perform the univariate comparison. Report the outputing statistic and p-value.
- Before you run the statistic test, determine the data type and check the missingness of two columns. In particular, report the number of NA values for each variable.

```python
# Write your own code
# ptsd_df.columns
# Figure out type for each variable
print(ptsd_df["pcl5week_score.completion"].dtype)
print()
print(ptsd_df["pcl5month_score.3_month_follow_up"].dtype)

# Determine the missingness for these columns
print(ptsd_df["pcl5week_score.completion"].isna().sum()) # N=27 are NA
print(ptsd_df["pcl5month_score.3_month_follow_up"].isna().sum()) # N=251 are NA

# assign vars to pcl_completion and pcl_3month (because I don't want to change the original vars when I drop missing values later on)
pcl_completion = ptsd_df["pcl5week_score.completion"]
pcl_3month = ptsd_df["pcl5month_score.3_month_follow_up"]

pcl5_data = pd.DataFrame({
    "pcl5week_score.completion": pcl_completion,
    "pcl5month_score.3_month_follow_up": pcl_3month
})

# generate summary statistics
def summary_stats(series):
    return {
        "mean": series.mean(),
        "std": series.std(),
        "median": series.median(),
        "Q1": series.quantile(0.25),
        "Q3": series.quantile(0.75)
    }

summary_table = pcl5_data.apply(summary_stats).T
print(summary_table)

# next, I will compute differences for paired test
# to do this, I need the same number of rows for each var
paired_data = pcl5_data.dropna(subset=["pcl5week_score.completion", "pcl5month_score.3_month_follow_up"])

diff_clean = paired_data["pcl5week_score.completion"] - paired_data["pcl5month_score.3_month_follow_up"]

shapiro_test = stats.shapiro(diff_clean)
print(f"\nShapiro-Wilk test for differences: W={shapiro_test.statistic:.3f}, p={shapiro_test.pvalue:.3f}")

if shapiro_test.pvalue > 0.05:
    t_stat, p_val = stats.ttest_rel(paired_data["pcl5week_score.completion"],
                                    paired_data["pcl5month_score.3_month_follow_up"])
    print(f"Paired t-test: t = {t_stat:.3f}, p = {p_val:.3f}")
else:
    w_stat, p_val = stats.wilcoxon(paired_data["pcl5week_score.completion"],
                                   paired_data["pcl5month_score.3_month_follow_up"])
    print(f"Wilcoxon signed-rank test: W = {w_stat:.3f}, p = {p_val:.3f}")
```

### Q2.2. Multiple Linear Regression (7')

<font size='4'>

- Select columns specified in the following code chunk and create a subset dataset named `ptsd_sub_df`.
- Fit a linear regression to examine the association between `caps_intake` (continuous outcome) and the remaining covariates (as predictors) using `ptsd_sub_df`.
    - Note that all covariates ending with `_code` are categorical variables.
- Use the instruction here to write the formula for linear regression in Python.
    - https://www.statsmodels.org/stable/example_formulas.html
- Report the output page including R2, adjusted R2, and parameter estimates, SE, 95% confidence intervals, and p-values.
- Provide a brief interpretation for all significant predictors (p<0.05) excluding the intercept.
- Relevant label information includes:
    - `employment_code`: 1: Employed, 2: Unemployed, 3: Retired, 4: Disabled/Unable to work, 5: Student, 6: Other.
    - `rank_code`: 1. Enlisted, 2: Officer, 3: Other

```python
# The following column names are used for linear regression.
# Do not delete.
ptsd_sub_df = ptsd_df[['caps_intake', 'age_iop', 'gender_code', 'sexualorient_code', 'race_code', 'ethnicity_code',
                      'education_code', 'employment_code',
                      'rank_code', 'branch_code', 'mdd_code', 'ctq_total_score', 'sexual_trauma', 'sud_code']].copy()
```

```python
# Write your own code
cat_vars = [col for col in ptsd_sub_df.columns if col.endswith('_code')]
# convert categorical vars to 'category' dtype
for col in cat_vars:
    ptsd_sub_df[col] = ptsd_sub_df[col].astype('category')
# identify predictor var
predictors = ptsd_sub_df.columns.drop('caps_intake')
formula_terms = []
for var in predictors:
    if var in cat_vars:
        formula_terms.append(f'C({var})')
    else:
        formula_terms.append(var)

formula = 'caps_intake ~ ' + ' + '.join(formula_terms)

# fit the linear model
model = smf.ols(formula=formula, data=ptsd_sub_df).fit()
print(model.summary())
```

```python
summary_df = pd.DataFrame({
    "Estimate": model.params,
    "Std_Error": model.bse,
    "CI_lower": model.conf_int()[0],
    "CI_upper": model.conf_int()[1],
    "p_value": model.pvalues
})
print(summary_df)
print()
```

Gender (see reference group) is associated with a -1.558164 decrease in caps_intake, assuming all other predictors are held constant  (p-value < 0.001). Similarly, for age, each additional year of age is associated with an average decrease of -0.038991 in caps_intake, assuming all other predictors are held constant (p-value < 0.001)...Other statistically significant variables of caps_intake (aside from gender and age) are sexual orientation, race, ethnicity, employment, rank code, branch code, mdd code, age, ctq_total_score, and sexual_trauma. These predictors are associated with increases / decreases in caps_intake assuming all other predictors are held constant. See estimates and corresponding p values in the output above.
