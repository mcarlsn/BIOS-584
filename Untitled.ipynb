```python
import os
import scipy.io as sio
from self_py_fun.HW10Fun import *
from sklearn.linear_model import LogisticRegression as LR
from sklearn.svm import SVC
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler
```

```python
# Global constants:
np.random.seed(100)
bp_low = 0.5
bp_upp = 6
electrode_num = 16
```

```python
parent_dir = '/Users/madelyncarlson/Documents/GitHub/BIOS-584'
parent_data_dir = '{}/data'.format(parent_dir)
time_index = np.linspace(0, 800, 25)
electrode_name_ls = ['F3', 'Fz', 'F4', 'T7', 'C3', 'Cz', 'C4', 'T8', 'CP3', 'CP4', 'P3', 'Pz', 'P4', 'PO7', 'PO8', 'Oz']
subject_name = 'K114'

# create a new folder called K114
subject_dir = '{}/{}'.format(parent_dir, subject_name)
if not os.path.exists(subject_dir):
    os.mkdir(subject_dir)

char_trn = 'THE0QUICK0BROWN0FOX'
char_trn_size = len(char_trn)
```

```python
trn_data_name = '{}_001_BCI_TRN_Truncated_Data_{}_{}'.format(subject_name, bp_low, bp_upp)
trn_data_dir = '{}/{}.mat'.format(parent_data_dir, trn_data_name)
eeg_trn_obj = sio.loadmat(trn_data_dir)
```

```python
print(eeg_trn_obj.keys())
eeg_trn_signal = eeg_trn_obj['Signal']
print(eeg_trn_signal.shape) # 3420, 400
eeg_trn_type = eeg_trn_obj['Type']
print(eeg_trn_type.shape) # 3420, 1
eeg_trn_type = np.squeeze(eeg_trn_type, axis=1)
```

```python
# Step 1.2: FRT dataset
# You should be able to obtain relevant data files: eeg_frt_signal and eeg_frt_type
frt_data_name = '{}_001_BCI_FRT_Truncated_Data_{}_{}'.format(subject_name, bp_low, bp_upp)
frt_data_dir = '{}/{}.mat'.format(parent_data_dir, frt_data_name)
eeg_frt_obj = sio.loadmat(frt_data_dir)
print(eeg_frt_obj.keys())

eeg_frt_signal = eeg_frt_obj['Signal']
print(eeg_frt_signal.shape) # (1296, 400) 
eeg_frt_type = eeg_frt_obj['Type']
print(eeg_frt_type.shape) # (1296, 1)
eeg_frt_type = np.squeeze(eeg_frt_type, axis=1)
```

```python
# Step 2: Fit classification models. You will try the following methods:
# Logistic Regression
# Linear Discriminant Analysis
# Support Vector Machine (sometimes called support vector classification)
# ...You do not need to modify the parameters of each classifier except for LogisticRegression: set max_iter=1000
```

```python
# instantiate models
logistic_obj = LR(max_iter=1000) # logistic regression  
lda_obj = LDA() # LDA 
svc_obj = SVC(probability=True) # SVM

# train models 
logistic_obj.fit(eeg_trn_signal, eeg_trn_type) # logistic regression  
lda_obj.fit(eeg_trn_signal, eeg_trn_type) # LDA
svc_obj.fit(eeg_trn_signal, eeg_trn_type) # SVM

# predict (FRT data) 
log_reg_preds = logistic_obj.predict(eeg_frt_signal) # logistic regression  
lda_preds = lda_obj.predict(eeg_frt_signal) # LDA
svc_preds = svc_obj.predict(eeg_frt_signal) # SVM

y_preds_three_method = np.stack([log_reg_preds, lda_preds, svc_preds], axis=0).T # stacking like we did in week 13
print(y_preds_three_method[:5, :])

log_reg_probs = logistic_obj.predict_proba(eeg_frt_signal)
print(log_reg_probs[:5, :])
print()

lda_probs = lda_obj.predict_proba(eeg_frt_signal)
print(lda_probs[:5, :])
print()

svc_probs = svc_obj.predict_proba(eeg_frt_signal)
print(svc_probs[:5, :])
print()
```

```python
# Step 3: Evaluate model performance on both TRN and FRT files
# store model predictions in a dictionary 
# TRN (can check for overfitting) 
log_reg_trn_preds = logistic_obj.predict(eeg_trn_signal)
lda_trn_preds = lda_obj.predict(eeg_trn_signal)
svc_trn_preds = svc_obj.predict(eeg_trn_signal)

model_preds_trn = {
    "Logistic Regression (TRN)": log_reg_trn_preds,
    "Linear Discriminant Analysis (TRN)": lda_trn_preds,
    "Support Vector Machine (TRN)": svc_trn_preds
}
for model, preds in model_preds_trn.items():
    print(f"{model} Results (TRN):")
    print(classification_report(eeg_trn_type, preds))
    print("\n")

# Comments on TRN output: Overall, models fit the TRN data. Logistic regression fits the training data very well;
# LDA performs similarly to logistic regression. Slightly lower F1 for class 1, but still strong;
# SVM has high precision but low recall.

# FRT 
model_preds = {
    "Logistic Regression": log_reg_preds,
    "Linear Discriminant Analysis": lda_preds,
    "Support Vector Machine": svc_preds
}
for model, preds in model_preds.items():
    print(f"{model} Results:")
    print(classification_report(eeg_frt_type, preds))
    print("\n")

# Comments on FRT output: Logistic regression still generalizes fairly well, but accuracy has dropped; 
# LDA generalized well on FRT data. It is reliable with balanced precision and recall;
# SVM is conservative on testing data. Still shows high precision but low recall. 
```

```python
# Step 3.1: Prediction accuracy on TRN files. We will compute the probability of each stimulus with .predict_proba()
# ...before we convert the stimulus-level probability to character-level probability. 
# You are asked to generate stimulus-level probability for each method on TRN files, denoted as logistic_y_trn, lda_y_trn, and svm_y_trn
scaler_obj = StandardScaler()
eeg_trn_signal_scaled = scaler_obj.fit_transform(eeg_trn_signal) 

logistic_y_trn = logistic_obj.predict_proba(eeg_trn_signal_scaled)
print("Logistic regression:", logistic_y_trn.shape)
print() 

lda_y_trn = lda_obj.predict_proba(eeg_trn_signal_scaled)
print("LDA:", lda_y_trn.shape)
print() 

svc_y_trn = svc_obj.predict_proba(eeg_trn_signal_scaled)
print("SVM:", svc_y_trn.shape)
print() 

logistic_y_trn_class1 = logistic_y_trn[:, 1]
print("Logistic regression:", logistic_y_trn_class1[:5]) # 0.0062 is very low probability for first trial
print() 

lda_y_trn_class1 = lda_y_trn[:, 1]
print("LDA:", lda_y_trn_class1[:5]) # slightly higher confidence for first trial
print() 

svc_y_trn_class1 = svc_y_trn[:, 1]
print("SVM:", svc_y_trn_class1[:5]) # very low confidence for first trial
print() 
```

```python
# Step 3.2: Prediction accuracy on FRT files
# Similarly, you are asked to generate stimulus-level probability for each method on FRT files, denoted as logistic_y_frt, lda_y_frt, and svm_y_frt.
eeg_frt_signal_scaled = scaler_obj.fit_transform(eeg_frt_signal) 

logistic_y_frt = logistic_obj.predict_proba(eeg_frt_signal_scaled)
print("Logistic regression shape:", logistic_y_frt.shape)
print() 
lda_y_frt = lda_obj.predict_proba(eeg_frt_signal_scaled)
print("LDA shape:", lda_y_frt.shape)
print() 
svc_y_frt = svc_obj.predict_proba(eeg_frt_signal_scaled)
print("SVM shape:", svc_y_frt.shape)
print() 

logistic_y_frt_class1 = logistic_y_frt[:, 1]
print("Logistic regression:", logistic_y_frt_class1[:5])
print() # observations: slightly lower than LDA. Moderate confidence. 

lda_y_frt_class1 = lda_y_frt[:, 1]
print("LDA:", lda_y_frt_class1[:5])
print() # observations: more confident 

svc_y_frt_class1 = svc_y_frt[:, 1]
print("SVM:", svc_y_frt_class1[:5])
print() # observations: high precision but low recall 
```
